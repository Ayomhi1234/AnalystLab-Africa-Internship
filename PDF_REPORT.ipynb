EXECUTIVE SUMMARY

This report presents a comprehensive analysis of advanced machine learning 
models applied to two datasets: House Price Prediction (Regression) and Titanic 
Survival (Classification). The objective was to build multiple models, optimize 
them through hyperparameter tuning, and identify the best-performing approach

4. EVALUATION METRICS
================================================================================

4.1 Regression Metrics (House Prices)

Mean Absolute Error (MAE)
- Definition: Average absolute difference between predicted and actual values
- Range: 0 to infinity (lower is better)
- Interpretation: Average prediction error in same units as target
- Advantage: Intuitive and not penalizing large errors heavily
- Unit: Currency (e.g., dollars)

Mean Squared Error (MSE)
- Definition: Average squared difference between predicted and actual values
- Range: 0 to infinity (lower is better)
- Interpretation: Penalizes larger errors more heavily
- Advantage: Mathematically convenient for optimization
- Disadvantage: Difficult to interpret (different units than target)

Root Mean Squared Error (RMSE)
- Definition: Square root of MSE
- Range: 0 to infinity (lower is better)
- Interpretation: Average error in same units as target
- Advantage: Interpretable and sensitive to large errors
- Most Used: Primary metric for regression comparison

R² Score (Coefficient of Determination)
- Definition: Proportion of variance in target explained by model
- Range: 0 to 1 (higher is better)
- Interpretation: 0.66 means model explains 66% of price variation
- Advantage: Scale-independent, easy comparison across datasets
- Limitation: Does not indicate bias in predictions


4.2 Classification Metrics (Titanic)

Accuracy
- Definition: Percentage of correct predictions
- Formula: (True Positives + True Negatives) / Total Predictions
- Range: 0 to 1 (higher is better)
- Interpretation: Overall correctness of model
- Limitation: Misleading with imbalanced classes

Precision
- Definition: Percentage of positive predictions that are correct
- Formula: True Positives / (True Positives + False Positives)
- Range: 0 to 1 (higher is better)
- Interpretation: If model predicts survived, how often correct?
- Use Case: When false positives are costly

Recall
- Definition: Percentage of actual positives correctly identified
- Formula: True Positives / (True Positives + False Negatives)
- Range: 0 to 1 (higher is better)
- Interpretation: How many actual survivors did model catch?
- Use Case: When false negatives are costly

F1-Score
- Definition: Harmonic mean of Precision and Recall
- Formula: 2 × (Precision × Recall) / (Precision + Recall)
- Range: 0 to 1 (higher is better)
- Interpretation: Balanced measure of precision and recall
- Use Case: When both false positives and negatives matter

ROC-AUC Score
- Definition: Area under Receiver Operating Characteristic curve
- Range: 0 to 1 (higher is better)
- Interpretation: Model's discrimination ability across all thresholds
- 0.5 = Random classifier, 1.0 = Perfect classifier
- Advantage: Threshold-independent, handles class imbalance well

6.1 Algorithm Effectiveness

Finding 1: Ensemble Methods Outperform Single Trees
- Gradient Boosting achieved best regression performance (R² = 0.66)
- Random Forest classifiers ranked among top performers
- Decision Trees consistently underperformed, indicating overfitting risk
- Implication: Combining multiple models through averaging or boosting captures 
  patterns better than single model approaches

Finding 2: Gradient Boosting Excels at Regression
- Best R² Score (0.66) and RMSE (1.3M) for house prices
- Iterative error correction captures complex non-linear price relationships
- Sequential building of trees leverages residual patterns
- Recommendation: Use Gradient Boosting for similar regression problems

Finding 3: Random Forest Effective for Classification
- Tuned Random Forest achieved best accuracy (81.56%)
- ROC-AUC of 0.7969 indicates excellent discrimination ability
- Balanced approach between bias and variance
- Recommendation: Use Random Forest as primary classifier for similar problems

Titanic Survival:

Finding 5: Feature Importance Aligns with Domain Knowledge

- Implication: Models are interpretable and trustworthy for decision-making


6.3 Hyperparameter Tuning Impact

Finding 6: Tuning Yields Modest Accuracy Gains but Improves Precision

House Prices:
- Tuned Random Forest R² = 0.59 vs Untuned = 0.61
- Tuning did NOT improve performance
- Untuned Gradient Boosting still superior
- Implication: Algorithm choice > parameter optimization for this dataset

Titanic:
- Tuned Random Forest Accuracy = 0.8156 vs Untuned = 0.8044 (improvement of 1.1%)
- Precision improved from 0.7671 to 0.8360 (8.9% increase)
- Recall decreased from 0.7567 to 0.6891 (trade-off)
- Implication: Tuning improved precision at cost of recall

Finding 7: GridSearchCV Testing 36 Combinations Efficient
- Systematic testing of parameter combinations
- Found optimal parameters for each dataset
- Trade-offs visible (e.g., precision vs recall)
- Implication: Structured tuning more reliable than random guessing


6.4 Model Robustness

Finding 8: Multiple Evaluation Metrics Essential
- Single metrics (accuracy alone) insufficient
- Precision-recall trade-off critical for classification
- Ensemble methods provide consistent performance across metrics
- Implication: Use multiple metrics for complete performance picture

Finding 9: Performance Consistency Across Ensemble Methods
- Random Forest, Gradient Boosting, Tuned Random Forest all achieve high scores
